```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import load_model
```

```python
# Load the model
model = load_model('cardio_model.h5')
```

```python
# Define the same feature names as in training
feature_names = ["age", "gender", "chestpain", "restingBP", "serumcholestrol",
                 "fastingbloodsugar", "restingrelectro", "maxheartrate",
                 "exerciseangia", "oldpeak", "slope", "noofmajorvessels"]
```

```python
# Prepare new data for prediction
new_data = pd.DataFrame({
    "age": [45],
    "gender": [1],
    "chestpain": [3],
    "restingBP": [120],
    "serumcholestrol": [230],
    "fastingbloodsugar": [0],
    "restingrelectro": [1],
    "maxheartrate": [150],
    "exerciseangia": [0],
    "oldpeak": [1.5],
    "slope": [2],
    "noofmajorvessels": [0]
})
```

```python
# Standardize the new data
scaler = StandardScaler()
```

```python
# Fit the scaler on training data features or use the saved scaler from training
# For simplicity, use a similar range of values as your training data during scaling
scaled_data = scaler.fit_transform(new_data)
```

```python
# Make predictions
predictions = model.predict(scaled_data)
```

```python
# Output prediction
predicted_classes = (predictions > 0.5).astype(int)
print(f"Predicted probability: {predictions[0][0]:.4f}")
print(f"Predicted class (1=Positive, 0=Negative): {predicted_classes[0][0]}")
```
